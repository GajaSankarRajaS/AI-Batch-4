```python
from transformers import pipeline

# Sentimental Analysis:
pipeline = pipeline("sentiment-analysis")
pipeline
```

```python
inputs = [
    "I am happy",
    "It's frustrating"
]

# Inference:
results = pipeline(inputs)
for text, result in zip(inputs, results):
  print(f"Text: {text}, Label: {result['label']}, Score: {result['score']:.2f}")
```

```python
# Classification:
from transformers import pipeline
classifier = pipeline("zero-shot-classification")

outputs = classifier(
    "Huggingface is a good website to learn AI/ML",
    candidate_labels = ["Education","Music"],
    hypothesis_template = "this example is about {}"
    )
outputs
```
