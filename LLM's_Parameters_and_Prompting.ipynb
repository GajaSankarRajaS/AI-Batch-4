```python
from google.colab import userdata
API_KEY = userdata.get('GROQ_TOKEN')
```

```python
import requests
import os

api_key = os.environ.get("GROQ_API_KEY")
url = "https://api.groq.com/openai/v1/models"

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

response = requests.get(url, headers=headers)

for data in response.json()['data']:
  print(data)
print()
```

```python
# Model Name:
model = "openai/gpt-oss-120b"
# Model Intialization:
from langchain_groq import ChatGroq

model = ChatGroq(
    model=model,
    api_key=API_KEY,
    temperature=0.0 # 0-1
)
```

```python
# Profile Inspect
model.profile
```

```python
# LLM Usage
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate(
    input_variables=["place"],
    template="Give me 3 interesting facts about a {place}.",
)

# Parser:
parser = StrOutputParser()

# Chain the process:
chain = prompt | model | parser

# Gettig the response:
response = chain.invoke({"place":"Tamil Nadu"})
print(response)
```

```python
# Few/One Shot Prompt
from langchain_core.prompts import ChatPromptTemplate, FewShotChatMessagePromptTemplate

# Step 1: Define your few-shot training examples
examples = [
    {"input": "The battery dies in 2 hours.", "output": "Negative | Battery Life"},
    {"input": "Customer support refunded me immediately.", "output": "Positive | Customer Service"},
    {"input": "The screen size is 6 inches.", "output": "Neutral | Product Specifications"}
]

# Step 2: Define how each individual example should be formatted
example_prompt = ChatPromptTemplate.from_messages([
    ("human", "{input}"),
    ("ai", "{output}")
])

# Step 3: Package the examples into a FewShotChatMessagePromptTemplate
few_shot_template = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples
)

# Step 4: Assemble the final complete prompt template
final_prompt = ChatPromptTemplate.from_messages([
    ("system", "Classify the sentiment and category of the customer review using the pattern shown below."),
    few_shot_template,
    ("human", "{input}")
])

# Step 5: Construct and execute the chain
few_shot_chain = final_prompt | model | parser

response = few_shot_chain.invoke({"input": "The camera lens is blurry and won't focus."})
print(response)
```
